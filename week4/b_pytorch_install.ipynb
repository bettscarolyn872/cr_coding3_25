## Step 0: Check Your Python Environment

First, let's see what Python environment you're using:

## Step 1: Check Existing Installations

Let's see what's already installed:

````python
import subprocess
import importlib

# Function to check if a package is installed
def check_package(package_name):
    try:
        importlib.import_module(package_name)
        return True, "Installed"
    except ImportError:
        return False, "Not Installed"

# Check required packages
packages_to_check = [
    ('torch', 'torch'),
    ('torchvision', 'torchvision'), 
    ('numpy', 'numpy'),
    ('matplotlib', 'matplotlib'),
    ('PIL', 'Pillow'),
    ('tqdm', 'tqdm')
]

print("=== Package Status ===")
for import_name, pip_name in packages_to_check:
    installed, status = check_package(import_name)
    print(f"{pip_name:15} : {status}")
    
# Check PyTorch details if installed
try:
    import torch
    print(f"\nPyTorch Version: {torch.__version__}")
    print(f"CUDA Available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"CUDA Version: {torch.version.cuda}")
except:
    print("\nPyTorch not installed yet.")
````

## Step 2: Choose Your Installation Method

Based on your system, choose ONE of the following methods:

### Option A: CPU-Only (Recommended for Beginners)
- Smaller download size
- Works on all systems
- Perfect for MNIST

### Option B: GPU with CUDA (If you have NVIDIA GPU)
- Faster training for large models
- Requires NVIDIA GPU
- Larger download size

````python
# Check if you have NVIDIA GPU
import subprocess

print("=== GPU Detection ===")
try:
    # Try to run nvidia-smi
    result = subprocess.run(['nvidia-smi'], capture_output=True, text=True)
    if result.returncode == 0:
        print("NVIDIA GPU detected!")
        print("You can use either CPU or GPU version.")
        # Show first few lines of nvidia-smi output
        lines = result.stdout.split('\n')[:8]
        print("\nGPU Info:")
        for line in lines:
            print(line)
    else:
        print("No NVIDIA GPU detected.")
        print("Use CPU-only version.")
except:
    print("nvidia-smi not found. No NVIDIA GPU or drivers not installed.")
    print("Use CPU-only version.")
````

## Step 3: Install PyTorch (CPU Version)

**Run this cell ONLY if you want CPU version:**
(Both Windows and Mac users can run on CPU)

````python
# OPTION A: CPU-ONLY INSTALLATION
print("Installing PyTorch CPU version...")
print("This may take a few minutes...\n")

import subprocess
import sys

# Install command
cmd = [sys.executable, '-m', 'pip', 'install', 
       'torch', 'torchvision', 'torchaudio', 
       '--index-url', 'https://download.pytorch.org/whl/cpu']

# Run installation
result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode == 0:
    print("PyTorch CPU version installed successfully!")
else:
    print("Installation failed. Error:")
    print(result.stderr)
````

## Step 3B: Install PyTorch (GPU Version)

**Run this cell ONLY if you have NVIDIA GPU and want GPU support:**

(You won't have this is you are using macOS. You will see CUDA later on, ignore this as well since it's not available for the Mac users.)

````python
# OPTION B: GPU INSTALLATION
print("Installing PyTorch GPU version...")
print("This will download ~2GB and may take several minutes...\n")

import subprocess
import sys

# For CUDA 11.8 (most common)
cmd = [sys.executable, '-m', 'pip', 'install', 
       'torch', 'torchvision', 'torchaudio',
       '--index-url', 'https://download.pytorch.org/whl/cu118']

# Run installation  
result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode == 0:
    print("PyTorch GPU version installed successfully!")
else:
    print("Installation failed. Error:")
    print(result.stderr)
    print("\nTry the CPU version instead (Step 3A)")
````

## Step 4: Install Other Required Packages

````python
# Install other required packages
print("Installing additional packages...")

packages = ['numpy', 'matplotlib', 'pillow', 'tqdm', 'jupyter', 'ipykernel']

for package in packages:
    print(f"\nInstalling {package}...")
    cmd = [sys.executable, '-m', 'pip', 'install', package]
    result = subprocess.run(cmd, capture_output=True, text=True)
    
    if result.returncode == 0:
        print(f"{package} installed")
    else:
        print(f"Failed to install {package}")
        print(result.stderr[:200])  # Show first 200 chars of error
````

## Troubleshooting Guide

### Common Issues and Solutions:

#### 1. **"No module named torch" after installation**
- **Solution**: Restart the kernel (very important!)
- In VS Code: Click the restart button (↻) or Kernel → Restart

#### 2. **"Permission denied" errors**
- **Solution**: Add `--user` flag:
  ```bash
  pip install --user torch torchvision
  ```

#### 3. **Wrong Python/pip version**
- **Solution**: Use python -m pip instead of just pip:
  ```bash
  python -m pip install torch torchvision
  ```

#### 4. **Kernel keeps crashing**
- **Solution**: Install in terminal instead:
  ```bash
  # Open VS Code terminal (Terminal → New Terminal)
  pip install torch torchvision matplotlib numpy
  ```

#### 5. **SSL Certificate errors**
- **Solution**: Use trusted host flag:
  ```bash
  pip install --trusted-host pypi.org --trusted-host files.pythonhosted.org torch
  ```

#### 6. **If all else fails, start with a new conda or environment**
