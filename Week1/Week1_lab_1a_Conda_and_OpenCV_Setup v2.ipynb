## Part 1: Install Conda (One-Time Setup)

**What is conda?**
Conda is a powerful package and environment manager that handles Python versions, packages, and dependencies automatically. It's widely used in data science and machine learning.

If you already have Anaconda or Miniconda installed, skip to Part 2.

### Option A: Install Miniconda (Recommended - Lighter)
1. Go to: https://docs.conda.io/en/latest/miniconda.html
2. Download the installer for your operating system
3. Run the installer and follow the prompts
4. **Important:** Check the box to "Add Conda to PATH" (or allow it to modify your shell configuration)

### Option B: Install Anaconda (Includes more packages)
1. Go to: https://www.anaconda.com/download
2. Download the installer for your operating system
3. Run the installer and follow the prompts

### Verify Installation
After installation, **close and reopen your terminal**, then run:
```bash
conda --version
```

You should see something like: `conda 23.x.x`

## Part 2: Create the Course Environment

Open your terminal/command prompt and run these commands:

### Step 1: Create a new Conda environment named 'coding3'

```bash
conda create -n coding3 python=3.10
```

When prompted, type `y` and press Enter.

**What this does:**
- Creates an isolated environment called `coding3`
- Installs Python 3.10 in this environment
- Keeps this course's packages separate from other projects

### Step 2: Activate the environment

```bash
conda activate coding3
```

You should see `(coding3)` appear at the beginning of your terminal prompt.

**Note:** You'll need to activate this environment every time you start a new terminal session for this course.

### Step 3: Install required packages

With the `coding3` environment activated, run:

```bash
conda install -c conda-forge opencv numpy jupyter ipykernel
```

Type `y` when prompted.

**What this installs:**
- `opencv` - Computer vision library
- `numpy` - Numerical computing library
- `jupyter` - Jupyter notebook interface
- `ipykernel` - Connects your environment to Jupyter notebooks

## Part 5: Useful Conda Commands

Here are some commands you'll use throughout the course:

### Activate/Deactivate Environment:
```bash
conda activate coding3      # Start working on the course
conda deactivate           # Switch back to base environment
```

### List Installed Packages:
```bash
conda list                 # See all packages in current environment
```

### Install Additional Packages:
```bash
conda install package_name           # Install from conda
pip install package_name             # Install from pip (if not in conda)
```

### List All Environments:
```bash
conda env list             # See all your conda environments
```

### Remove Environment (when course is done):
```bash
conda env remove -n coding3
```

## Troubleshooting

### Problem: "conda: command not found"
**Solution:** 
- Close and reopen your terminal after installing Conda
- Or run: `export PATH="$HOME/miniconda3/bin:$PATH"` (macOS/Linux)

### Problem: Kernel doesn't show in VS Code
**Solution:**
1. Restart VS Code
2. Install Python extension for VS Code if not already installed
3. Run: `conda install ipykernel` in the coding3 environment

### Problem: "PackagesNotFoundError"
**Solution:**
- Make sure you're using the conda-forge channel: `-c conda-forge`
- Or install with pip: `pip install opencv-python`

### Problem: Wrong Python version running
**Solution:**
- Check you selected the `coding3` kernel in VS Code (top-right)
- Run the verification cell above to confirm
