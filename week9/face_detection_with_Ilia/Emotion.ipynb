```python
from deepface import DeepFace

import os
os.environ["OPENCV_VIDEOIO_MSMF_ENABLE_HW_TRANSFORMS"] = "0"
import cv2
```

```python
# font
font = cv2.FONT_HERSHEY_SIMPLEX

# fontScale
fontScale = 1
 
# Blue color in BGR
color = (255, 0, 0)

# Line thickness of 2 px
thickness = 2

def annotate_image(image, region, emotion):
    image = cv2.rectangle(image, (region['x'], region['y']), 
                          (region['x']+region['w'], region['y']+region['h']), (255, 0, 0), 2)
    image = cv2.putText(image, emotion, (region['x'], region['y']), font, 
                        fontScale, color, thickness, cv2.LINE_AA)
    return image
```

```python
# cam.release()
cam = cv2.VideoCapture(0, cv2.CAP_DSHOW)

while True:
    ret, frame = cam.read()

    objs = DeepFace.analyze(img_path = frame, actions = ['age', 'emotion'], enforce_detection=False)
    embedding_objs = DeepFace.represent(img_path = frame, enforce_detection=False)

    # embedding_objs_memory

    if len(objs) > 0:
        for region in objs:
            frame = annotate_image(frame, region['region'], region['dominant_emotion'])
    
    cv2.imshow("0", frame)
    key = cv2.waitKey(1)

    if key == 27:
        break
        
cam.release()
cv2.destroyAllWindows()
```

```python
from numpy import dot
from numpy.linalg import norm

a = embedding_objs[0]['embedding']
b = embedding_objs[1]['embedding']

cos_sim = dot(a, b)/(norm(a)*norm(b))

print(cos_sim)
```
