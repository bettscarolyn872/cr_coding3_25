## Part 1: OpenCV Basics - Understanding Images

In OpenCV, images are represented as NumPy arrays with dimensions:
- **Grayscale:** (height, width)
- **Color (BGR):** (height, width, 3)

### Exercise 1.1: Create a Simple Image

```python
import cv2
import numpy as np

# Create a blank black image (100x100 pixels)
# np.zeros creates an array filled with zeros
# dtype=np.uint8 means 8-bit unsigned integer (0-255 range)
black_image = np.zeros((100, 100, 3), dtype=np.uint8)

print(f"Image shape: {black_image.shape}")
print(f"Image data type: {black_image.dtype}")
print(f"Image dimensions: Height={black_image.shape[0]}, Width={black_image.shape[1]}, Channels={black_image.shape[2]}")
```

### Exercise 1.2: Create Colored Images

**⚠IMPORTANT NOTE: BGR vs. RGB⚠**

OpenCV's default color format is BGR (Blue, Green, Red), not the standard RGB. 
When you set a pixel to (255, 0, 0), you are setting the Blue channel to maximum. Keep this BGR order in mind for all color operations, including Exercise 4 (Your Design).

```python
# Create a red image
# Note: OpenCV uses BGR (Blue, Green, Red) format, not RGB!
red_image = np.zeros((100, 100, 3), dtype=np.uint8)
red_image[:, :] = (0, 0, 255)  # BGR format: (Blue=0, Green=0, Red=255)

# Create a green image
green_image = np.zeros((100, 100, 3), dtype=np.uint8)
green_image[:, :] = (0, 255, 0)  # BGR format

# Create a blue image
blue_image = np.zeros((100, 100, 3), dtype=np.uint8)
blue_image[:, :] = (255, 0, 0)  # BGR format

print("Created red, green, and blue images")
print(f"Red image pixel at (0,0): {red_image[0, 0]}")
```

### Exercise 1.3: Draw Shapes on Images

```python
# Create a blank canvas
canvas = np.zeros((400, 400, 3), dtype=np.uint8)

# Draw a rectangle
# cv2.rectangle(image, top_left_corner, bottom_right_corner, color, thickness)
cv2.rectangle(canvas, (50, 50), (150, 150), (0, 255, 0), 2)

# Draw a circle
# cv2.circle(image, center, radius, color, thickness)
cv2.circle(canvas, (250, 100), 50, (255, 0, 0), -1)  # -1 fills the circle

# Draw a line
# cv2.line(image, start_point, end_point, color, thickness)
cv2.line(canvas, (50, 250), (350, 250), (0, 0, 255), 3)

# Add text
# cv2.putText(image, text, position, font, scale, color, thickness)
cv2.putText(canvas, "Hello OpenCV!", (100, 350), 
            cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

print("Shapes drawn successfully!")
print("To see the image, we need to save or display it.")
```

### Exercise 1.4: Save Your First Image

```python
# Save the canvas image to a file
output_filename = "hello_opencv.png"
cv2.imwrite(output_filename, canvas)

print(f"✓ Image saved as '{output_filename}'")
print(f"Check your current directory to view the image!")
```

## Part 2: Loading and Manipulating Images

### Exercise 2.1: Load an Image

**Note:** For this exercise, you'll need an image file in your working directory. You can use the image we just created, or download a test image.

## Core Concept: Images are NumPy Arrays
When you load an image using `cv2.imread()`, the `loaded_image` variable is a high-performance NumPy N-dimensional array (`ndarray`). 

This is the most critical concept in OpenCV:

**Pixel Access (Exercise 2.2)**: When you use bracket notation like `loaded_image[100, 100]` to access a pixel, you are using NumPy's fast indexing/slicing feature, not standard, slower Python list indexing.

**Image Cropping (Exercise 3.3)**: The efficient slicing `loaded_image[50:200, 50:200]` is also a NumPy array slicing operation (called a "view" of the array), which is how OpenCV handles regions of interest (ROI).

**Efficiency**: This is why NumPy is required for this course. These vectorized operations are essential for the speed needed in Computer Vision tasks like webcam processing (Part 5).

```python
# Load the image we just created
# cv2.imread() returns a NumPy ndarray
loaded_image = cv2.imread("hello_opencv.png")

if loaded_image is not None:
    print("✓ Image loaded successfully!")
    print(f"Image shape: {loaded_image.shape}")
    print(f"Image size: {loaded_image.size} pixels")
    print(f"Image data type: {loaded_image.dtype}")
else:
    print("✗ Error: Could not load image")
    print("Make sure 'hello_opencv.png' exists in your current directory")
```

### Exercise 2.2: Image Properties and Pixel Access

```python
# Access individual pixel values
if loaded_image is not None:
    # Get pixel at position (100, 100)
    pixel = loaded_image[100, 100]
    print(f"Pixel at (100, 100): BGR = {pixel}")
    
    # Get dimensions
    height, width, channels = loaded_image.shape
    print(f"\nImage dimensions:")
    print(f"  Height: {height} pixels")
    print(f"  Width: {width} pixels")
    print(f"  Channels: {channels}")
    
    # Modify a pixel
    loaded_image[100, 100] = (255, 255, 255)  # Set to white
    print(f"\nModified pixel at (100, 100): BGR = {loaded_image[100, 100]}")
```

### Exercise 2.3: Color Space Conversion

```python
if loaded_image is not None:
    # Convert BGR to Grayscale
    gray_image = cv2.cvtColor(loaded_image, cv2.COLOR_BGR2GRAY)
    print(f"Grayscale image shape: {gray_image.shape}")
    
    # Convert BGR to HSV (Hue, Saturation, Value)
    hsv_image = cv2.cvtColor(loaded_image, cv2.COLOR_BGR2HSV)
    print(f"HSV image shape: {hsv_image.shape}")
    
    # Save the grayscale version
    cv2.imwrite("hello_opencv_gray.png", gray_image)
    print("\n✓ Grayscale image saved as 'hello_opencv_gray.png'")
```

## Part 3: Basic Image Operations

### Exercise 3.1: Image Resizing

```python
if loaded_image is not None:
    # Resize to specific dimensions
    resized = cv2.resize(loaded_image, (200, 200))
    print(f"Original size: {loaded_image.shape}")
    print(f"Resized size: {resized.shape}")
    
    # Resize by scale factor
    scale_percent = 50  # 50% of original size
    width = int(loaded_image.shape[1] * scale_percent / 100)
    height = int(loaded_image.shape[0] * scale_percent / 100)
    scaled = cv2.resize(loaded_image, (width, height))
    
    cv2.imwrite("hello_opencv_resized.png", resized)
    print("\n✓ Resized image saved")
```

### Exercise 3.2: Image Rotation

```python
if loaded_image is not None:
    # Get image dimensions
    height, width = loaded_image.shape[:2]
    
    # Calculate rotation matrix (rotate 45 degrees around center)
    center = (width // 2, height // 2)
    angle = 45
    scale = 1.0
    
    rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)
    rotated = cv2.warpAffine(loaded_image, rotation_matrix, (width, height))
    
    cv2.imwrite("hello_opencv_rotated.png", rotated)
    print("✓ Rotated image saved as 'hello_opencv_rotated.png'")
```

### Exercise 3.3: Image Cropping

```python
if loaded_image is not None:
    # Crop using array slicing: image[y1:y2, x1:x2]
    cropped = loaded_image[50:200, 50:200]
    
    print(f"Original shape: {loaded_image.shape}")
    print(f"Cropped shape: {cropped.shape}")
    
    cv2.imwrite("hello_opencv_cropped.png", cropped)
    print("\n✓ Cropped image saved")
```

## Part 5: Working with Webcam (Optional)

**Note:** This section captures a single frame from your webcam. Make sure you have a webcam connected.

```python
# Capture a single frame from webcam
# Note: This may not work in all notebook environments

# Initialize camera (0 is usually the default camera)
cap = cv2.VideoCapture(0)

# Manually adjust brightness
cap.set(cv2.CAP_PROP_BRIGHTNESS, 0.5)  # 0 to 1
cap.set(cv2.CAP_PROP_CONTRAST, 0.5)
cap.set(cv2.CAP_PROP_EXPOSURE, -5)

if cap.isOpened():
    # Capture a single frame
    ret, frame = cap.read()
    
    if ret:
        # Save the captured frame
        cv2.imwrite("webcam_capture.png", frame)
        print("✓ Webcam frame captured and saved as 'webcam_capture.png'")
        print(f"Frame shape: {frame.shape}")
    else:
        print("✗ Failed to capture frame")
    
    # Release the camera
    cap.release()
else:
    print("✗ Could not open webcam")
    print("This is normal if you're running in certain notebook environments")
```
